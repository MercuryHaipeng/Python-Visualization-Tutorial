### 第5章 第三方图表 pyecharts
- pyecharts 是一个基于ECharts的Python数据可视化工具，用于生成交互式的、动态图表，可以生成独立的网页格式图表，也可以嵌入 Jupyer Notebook、HTML 文件和 Web 应用中
- pyecharts 对方法的调用分为单独调用和链式调用。单独调用就是常规的一个接一个方法调用；而链式调用将所有需要调用的方法写在一个方法里，代码看上去更加简洁，也更易懂
- 定制图表主题：`class InitOpts(width:str='900px, height:str='500px', chart_id:Optional[str]=None, renderer:str=RenderType.CANVAS, page_title:str='Awesome-pyecharts', theme:str='white', bg_color:Optional[str]=None, js_host:str='', animation_opts:Unino[AnimationOpts, dict]=AnimationOpts(),)`
    - `width` 字符型，图表画布宽度，以像素为单位
    - `height` 字符型，图表画布高度，以像素为单位
    - `chart_id` 图表唯一 ID 标识，主要用于绘制多图表时区分每个子图表
    - `page_title` 字符型，网页标题
    - `theme` 字符型，图表主题风格，内置15种不同的主题风格
        - `ThemeType.WHITE` 默认主题
        - `ThemeType.DARK` 深色主题
        - `ThemeType.LIGHT` 浅色主题
        - `ThemeType.CHALK` 粉笔色主题
        - `ThemeType.ESSOS` 厄索斯大陆主题
        - `ThemeType.INFOGRAPHIC` 信息图主题
        - `ThemeType.MACARONS` 马卡龙主题
        - `ThemeType.PURPLE_PASSION` 紫色热烈主题
        - `ThemeType.ROMA` 罗马假日主题
        - `ThemeType.ROMANTIC` 浪漫主题
        - `ThemeType.SHINE` 闪耀主题
        - `ThemeType.VINTAGE` 葡萄酒主题
        - `ThemeType.WALDEN` 瓦尔登湖主题
        - `ThemeType.WESTEROS` 维斯特洛大陆主题
        - `ThemeType.WONDERLAND` 仙境主题
    - `bg_color` 字符型，图表背景颜色
